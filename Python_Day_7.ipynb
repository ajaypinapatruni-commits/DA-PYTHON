```python
#marks grade calculation
x,y,z=map(int,input("enter three subjects").split())
total=x+y+z
print(total)
avg=total/3
print(avg)
if x < 35 or y < 35 or z < 35:
  print("f")
elif avg>=90:
  print("A grade")
elif (avg>=75 and avg<=89):
  print("B grade")
elif (avg>=60 and avg<=74):
  print("C grade")
elif (avg>=40 and avg<=59):
  print("D grade")
else:
  print("F grade")



```

```python
units = int(input())
total = 0

if units <= 100:
    total = units * 2
elif units <= 200:
    total = 100*2 + (units - 100)*3
elif units <= 400:
    total = 100*2 + 100*3 + (units - 200)*5
else:
    total = 100*2 + 100*3 + 200*5 + (units - 400)*7

print(total)

if total > 2000:
    total = total * 1.10
    print(total)

print(f"{total:.2f}")
```
