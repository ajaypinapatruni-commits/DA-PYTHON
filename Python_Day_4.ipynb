```python
### Bitwise Operators: Names and Meanings

| Operator | Name | Meaning | Example (`a = 5` [101], `b = 3` [011]) |
| :---: | :--- | :--- | :--- |
| `&` | **AND** | Sets each bit to `1` if both bits are `1`. | `a & b` $\rightarrow$ `1` (`001`) |
| `\|` | **OR** | Sets each bit to `1` if at least one bit is `1`. | `a \| b` $\rightarrow$ `7` (`111`) |
| `^` | **XOR** (Exclusive OR) | Sets each bit to `1` only if one of the bits is `1` (but not both). | `a ^ b` $\rightarrow$ `6` (`110`) |
| `~` | **NOT** | Inverts all the bits (flips `1` to `0` and `0` to `1`). | `~a` $\rightarrow$ `-6` |
| `<<` | **Zero-fill left shift** | Shifts bits left by pushing zeros in from the right. | `a << 1` $\rightarrow$ `10` (`1010`) |
| `>>` | **Signed right shift** | Shifts bits right by pushing copies of the leftmost bit in from the left. | `a >> 1` $\rightarrow$ `2` (`010`) |
```
